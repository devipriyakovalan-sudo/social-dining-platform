```python
# 2)Write a python code to find the relation between Mid Term Exam (MTE) and Final
#Assessment Test (FAT). Apply linear regression to predict the FAT score given MTE
#score. (with and without package)

# Linear Regression Without Package

# MTE and FAT scores
x = [40, 50, 60, 70, 80]
y = [45, 55, 65, 75, 85]

n = len(x)

sum_x = sum(x)
sum_y = sum(y)
sum_xy = sum([x[i] * y[i] for i in range(n)])
sum_x2 = sum([i * i for i in x])

# Calculate slope (m)
m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)

# Calculate intercept (c)
c = (sum_y - m * sum_x) / n

print("Slope =", m)
print("Intercept =", c)

# Predict FAT score
mte = float(input("Enter MTE Score: "))
fat = m * mte + c

print("Predicted FAT Score =", fat)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# MTE and FAT scores
X = np.array([40, 50, 60, 70, 80]).reshape(-1, 1)
Y = np.array([45, 55, 65, 75, 85])

# Create model
model = LinearRegression()

# Train model
model.fit(X, Y)

print("Slope =", model.coef_[0])
print("Intercept =", model.intercept_)

# Predict FAT score
mte = float(input("Enter MTE Score: "))
prediction = model.predict([[mte]])

print("Predicted FAT Score =", prediction[0])
```
